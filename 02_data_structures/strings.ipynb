```python
#Anagram

def is_anagram(word1: str, word2: str)->bool:
    
    word1 = word1.lower().replace(" ", "")

    word2 = word2.lower().replace(" ", "")

    return sorted(word1) == sorted(word2)
```

```python
#Anagram Collections counter 

from collections import Counter

def is_anagram_collections(word1: str, word2: str) -> bool:
    word1 = word1.lower().replace("", " ")
    word2 = word2.lower().replace("", " ")

    return Counter(word1) == Counter(word2)


print(is_anagram("Roma", "Amor"))         # Resultado: True
print(is_anagram("Alergia", "Alegria"))   # Resultado: True
print(is_anagram("Gato", "Pato"))         # Resultado: False

print(is_anagram_collections("Roma", "Amor"))         # Resultado: True
print(is_anagram_collections("Alergia", "Alegria"))   # Resultado: True
print(is_anagram_collections("Gato", "Pato"))         # Resultado: False

```
