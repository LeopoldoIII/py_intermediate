## Section 2: Tuple Unpacking and Extended Unpacking

### 🎯 Objective
Use structural tuple unpacking to assign tuple components directly into distinct variables, and leverage extended unpacking `*rest` to capture variable-length sub-sequences.

---

### 📖 1. Code Sample
```python
data = ("Alice", 95, 88, 92, "Passed")
name, *scores, status = data
print(f"Name: {name}, Scores: {scores}, Status: {status}")
# Output: Name: Alice, Scores: [95, 88, 92], Status: Passed
```

## Section 3: Sorting Tuples by Specific Index Key

### 🎯 Objective
Sort collections of tuple records based on arbitrary column indices using Python's `sorted()` function with a custom `key=lambda x: x[index]` parameter.

---

### 📖 1. Code Sample
```python
employees = [("Alice", 75000), ("Bob", 60000), ("Charlie", 90000)]
# Sort by salary (index 1) descending:
sorted_by_salary = sorted(employees, key=lambda emp: emp[1], reverse=True)
print("Sorted by salary:", sorted_by_salary)
```
