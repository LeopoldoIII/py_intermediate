        groups[key].append(word)

## Section 3: collections.deque & Sliding Window Rate Limiting

### 🎯 Objective
Understand why double-ended queues (`deque`) are preferred over standard lists for FIFO queues: $O(1)$ constant time `popleft()` and `appendleft()`, compared to $O(n)$ for `list.pop(0)`.

---

### 📖 1. Code Sample
```python
from collections import deque

queue = deque(["task1", "task2"])
queue.append("task3")       # Push right: O(1)
first = queue.popleft()     # Pop left: O(1)
print(f"Processed: {first}, Remaining: {list(queue)}")
```

<summary>💡 Hint</summary>
Maintain a `deque`. For each timestamp `t`, `dq.append(t)`. While `dq[0] < t - window_size`, `dq.popleft()`. Append `len(dq)` to the result list.

<details>
<summary>🔍 Solution</summary>

```python
from collections import deque

def count_recent_calls(timestamps: list[int], window_size: int = 3000) -> list[int]:
    sliding_window = deque()
    results = []
    for t in timestamps:
        sliding_window.append(t)
        while sliding_window and sliding_window[0] < t - window_size:
            sliding_window.popleft()
        results.append(len(sliding_window))
    return results
```

## Section 4: collections.namedtuple & Geometric Distance

### 🎯 Objective
