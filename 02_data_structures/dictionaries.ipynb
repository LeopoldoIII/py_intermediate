# Dictionaries in Python

Dictionaries are mutable, ordered (since Python 3.7+) mappings of unique keys to values. They provide average $O(1)$ time complexity for lookups, insertions, and deletions via hash tables.

def get_nested_value(data: dict, keys: list[str], default: Any = None) -> Any:

            current = current[k]
        else:
            return default
    return current

## Section 2: Inverting Key-Value Mappings

### 🎯 Objective
Traverse dictionary items using `.items()` and invert key-value associations, grouping keys that share identical values.

---

### 📖 1. Code Sample
```python
grades = {"Alice": "A", "Bob": "B", "Charlie": "A"}
# Inverting mapping:
grouped = {}
for student, grade in grades.items():
    grouped.setdefault(grade, []).append(student)
print("Students grouped by grade:", grouped)
```

def invert_dictionary(mapping: dict[str, int]) -> dict[int, list[str]]:
    inverted: dict[int, list[str]] = {}
    for key, val in mapping.items():
        inverted.setdefault(val, []).append(key)
    return inverted
