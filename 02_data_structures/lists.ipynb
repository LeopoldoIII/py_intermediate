## Section 1: In-Place Mutation and Element Rotation

### 🎯 Objective
Understand how lists handle dynamic sizing and modifications. Learn the performance difference between $O(1)$ amortized operations (`append`, `pop`) and $O(n)$ operations (`insert`, `remove`).

---

### 📖 1. Code Sample
```python
fruits = ["banana", "cherry", "apple"]
fruits.append("lemon")         # O(1) append to end
fruits.insert(1, "blueberry")  # O(n) insert at index 1
removed_item = fruits.pop()    # O(1) pop last element
print("Fruits:", fruits)
print("Popped:", removed_item)
```

## Section 2: Extended Slicing and Memory-Safe Copying

### 🎯 Objective
Master Python slice syntax `[start:stop:step]` and avoid accidental mutation side effects by creating proper shallow copies (`.copy()`, `[:]`) instead of variable aliasing.

---

### 📖 1. Code Sample

<summary>🔍 Solution</summary>

```python
def split_camel_case(text: str) -> str:
    if not text:
        return ""
    characters = [
        (" " + char) if (char.isupper() and idx > 0) else char
        for idx, char in enumerate(text)
    ]
    return "".join(characters)
```
