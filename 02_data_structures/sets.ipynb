# Sets in Python

Sets are mutable, unordered collections of unique, hashable elements. They are optimized for $O(1)$ membership checking and mathematical set operations (union, intersection, difference, symmetric difference).

## Section 1: Set Uniqueness & Deduplication

### 🎯 Objective
Understand how sets enforce element uniqueness and how to remove duplicate items from collections while maintaining fast $O(1)$ lookups.

---

### 📖 1. Code Sample
```python
raw_ids = [101, 102, 101, 103, 102, 104]
unique_ids = set(raw_ids)
print("Unique IDs:", unique_ids)  # {101, 102, 103, 104}
print("Is 103 in set?", 103 in unique_ids)  # O(1) membership check: True
```

## Section 2: Mathematical Set Operations (Jaccard Similarity)

### 🎯 Objective
Master the 4 core mathematical set operators:
- **Union (`|`)**: Elements in either set.
- **Intersection (`&`)**: Elements in both sets.
- **Difference (`-`)**: Elements in first set but not second.
- **Symmetric Difference (`^`)**: Elements in exactly one set.

---

### 📖 1. Code Sample
```python
a = {1, 2, 3}
b = {2, 3, 4}
print("Union:", a | b)                 # {1, 2, 3, 4}
print("Intersection:", a & b)          # {2, 3}
print("Difference (a - b):", a - b)    # {1}
print("Symmetric Diff:", a ^ b)        # {1, 4}
```

    union = set_a | set_b
    if not union:

    intersection = set_a & set_b
    return len(intersection) / len(union)
