## Exercise 2: Count Set Bits (Hamming Weight)
Write a function `count_set_bits(n: int) -> int` that returns how many bits are set to `1` in the binary representation of a positive integer `n`.

```python
def count_set_bits(n: int) -> int:
    # Write your code here
    if hasattr(n, "bit_count"):
        return n.bit_count()
    return bin(n).count("1")

# Test
print("Set bits in 37:", count_set_bits(37)) # Expected: 3 (0b100101)
```

## Exercise 3: Byte Serialization and Deserialization
Take the user ID `user_id = 305419896` (which is `0x12345678` in hexadecimal):
1. Convert it into a 4-byte `bytes` object using Big-Endian byte order.
2. Reconstruct the integer from the generated `bytes` object and verify it matches the original.

```python
user_id = 305419896

# Write your code here
byte_data = user_id.to_bytes(4, byteorder="big")
reconstructed_id = int.from_bytes(byte_data, byteorder="big")

print("Bytes:", byte_data)
print("Reconstructed:", reconstructed_id)
print("Matches:", user_id == reconstructed_id)
```

## Exercise 4: Format Time Duration using divmod
Given a total duration of `7384` seconds, calculate the hours, minutes, and remaining seconds using `divmod()`, and format the result as `HH:MM:SS`.

```python
total_seconds = 7384

# Write your code here
hours, remainder = divmod(total_seconds, 3600)
minutes, seconds = divmod(remainder, 60)

formatted_time = f"{hours:02d}:{minutes:02d}:{seconds:02d}"
print("Formatted time:", formatted_time)  # Expected: 02:03:04
```

## Exercise 5: Fast Modular Exponentiation
Calculate $(7^{256}) \pmod{13}$ efficiently using Python's built-in `pow()` function.

```python
base = 7
exponent = 256
modulus = 13

# Write your code here
result = pow(base, exponent, modulus)
print("Result:", result)
```
