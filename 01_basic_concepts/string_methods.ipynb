# String Methods in Python
Strings in Python come with many built-in methods that allow you to easily perform checks and modifications. Here is a quick reference grouped by utility.

## 1. Content Checking (Booleans)
These methods return `True` or `False` depending on the string's content.

```python
s = "Hola123"

# isalpha(): True if ALL characters are alphabetic letters
print(f"isalpha: {s}", s.isalpha()) # False due to numbers
print("isalpha (letters only):", "Hola".isalpha())

# isdigit(): True if ALL characters are numbers
print("isdigit:", "12345".isdigit())

# isalnum(): True if characters are letters OR numbers (no spaces or special characters)
print("isalnum:", s.isalnum())

# islower() / isupper(): True if all letters are lowercase / uppercase
print("islower:", "hola".islower())
print("isupper:", "HOLA".isupper())

# isspace(): True if the string only contains whitespace characters
print("isspace:", "   ".isspace())
```

## 2. Case Transformation

```python
s = "hola Mundo"

# upper() and lower()
print("upper:", s.upper())
print("lower:", s.lower())

# capitalize(): Only the first letter of the entire string is capitalized
print("capitalize:", s.capitalize())

# title(): The first letter of EACH word is capitalized
print("title:", s.title())

# swapcase(): Inverts uppercase and lowercase
print("swapcase:", "HoLa".swapcase())
```

## 3. Search and Replace

```python
s = "Python is great and Python is fast"

# find(): Returns the index of the first occurrence. Returns -1 if not found
print("find:", s.find("is"))

# count(): Counts how many times a substring appears
print("count:", s.count("Python"))

# startswith() / endswith(): True if it starts or ends with the given substring
print("startswith:", s.startswith("Py"))
print("endswith:", s.endswith("slow"))

# replace(old, new): Replaces the text (you can pass a third parameter to limit occurrences)
print("replace:", s.replace("Python", "Java", 1)) # Replaces only 1 time
```

## 4. Split and Join (Lists and Strings)

```python
# split(): Splits the string into a list. By default, it splits by whitespace.
s = "apple,pear,grape"
fruits = s.split(",")
print("split:", fruits)

# join(): Joins a list (or iterable) of strings using the original string as a separator
separator = " - "
joined = separator.join(fruits)
print("join:", joined)

# partition(): Splits the string into a 3-element tuple (before separator, separator, after)
print("partition:", "user@email.com".partition("@"))
```

## 5. Whitespace Cleaning (Strip)

```python
s = "   text with extra spaces   "

# strip(): Removes whitespace at the beginning and end
print("strip:", f"\"{s.strip()}\"")

# lstrip() / rstrip(): Removes whitespace only on the left or right
print("lstrip:", f"\"{s.lstrip()}\"")
print("rstrip:", f"\"{s.rstrip()}\"")
```

## 6. Padding and Alignment

```python
s = "42"

# zfill(width): Pads with zeros on the left until reaching the width
print("zfill:", s.zfill(5)) # 00042

# ljust(width, fillchar) / rjust(width, fillchar): Justifies by adding characters
print("ljust:", s.ljust(10, "-"))
print("rjust:", s.rjust(10, "*"))

# center(width, fillchar): Centers the text
print("center:", "Title".center(30, "="))
```
