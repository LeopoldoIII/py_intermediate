### 📖 1. Code Sample

```python
for num in range(1, 10):
    if num % 2 != 0:
        odd_sum += num
print("Sum of odd numbers < 10:", odd_sum)  # 1 + 3 + 5 + 7 + 9 = 25
```

## Section 3: Reverse String using Loop Traversal

### 🎯 Objective
Traverse a string character-by-character from end to start using loops and indices, demonstrating manual string reversal without relying immediately on `[::-1]`.

---

### 📖 1. Code Sample
```python

reversed_chars = []
for i in range(len(text) - 1, -1, -1):
    reversed_chars.append(text[i])
print("Reversed:", "".join(reversed_chars))  # "nohtyP"
```

<details>

```python
def reverse_text_loop(text: str, ignore_spaces: bool = False) -> str:
    if ignore_spaces:
        text = text.replace(" ", "")
    result = []
    for char in reversed(text):
        result.append(char)
    return "".join(result)
```
