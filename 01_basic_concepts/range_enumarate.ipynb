## Section 2: Enumerate vs range(len()) Antipattern

### 🎯 Objective
Learn why using `for i in range(len(items)):` is an unpythonic antipattern, and how `enumerate(items, start=0)` yields clean `(index, value)` tuples directly.

---

### 📖 1. Code Sample
```python
fruits = ["apple", "banana", "cherry"]
# Idiomatic enumeration:
for index, fruit in enumerate(fruits, start=1):
    print(f"{index}. {fruit}")
```

    result = list(numbers)
    for i, num in enumerate(result):
        if num % 15 == 0:
            result[i] = "fizzbuzz"
        elif num % 3 == 0:
            result[i] = "fizz"
        elif num % 5 == 0:
            result[i] = "buzz"
    return result
