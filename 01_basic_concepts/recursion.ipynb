Understand the two essential requirements of every recursive function:
1. **Base Case**: The stopping condition that prevents infinite recursion (`RecursionError: maximum recursion depth exceeded`).
2. **Recursive Step**: Moving toward the base case by reducing the input problem.

---

### 📖 1. Code Sample
```python
# Countdown recursive function
def countdown(n: int) -> list[int]:
    if n <= 0:
        return []
    return [n] + countdown(n - 1)

print("Countdown:", countdown(4))  # [4, 3, 2, 1]
```
