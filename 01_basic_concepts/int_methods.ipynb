# Integer Methods in Python
Integers (`int`) in Python come with built-in methods and attributes that are useful for low-level operations, binary representation, byte serialization, and numerical duck typing.

> **Syntax Note:** When calling a method directly on an integer literal, surround the number with parentheses (e.g. `(42).bit_length()`) or assign it to a variable to prevent Python from parsing the dot as a decimal point.

## 1. Bit Representation and Counting
Python integers provide methods to inspect their binary representation.

```python
n = 37  # 37 in binary is 0b100101

# bit_length(): Number of bits necessary to represent the integer in binary
print(f"bit_length of {n}:", n.bit_length())  # 6

# bit_count(): Number of ones ('1's) in the binary representation
# Note: Available in Python 3.10+
try:
    print(f"bit_count of {n}:", n.bit_count())
except AttributeError:
    # Fallback for Python < 3.10
    print(f"bit_count (fallback) of {n}:", bin(n).count("1"))
```

## 2. Byte Conversion (to_bytes and from_bytes)
These methods allow converting integers to raw bytes and vice versa, essential for networking, file formats, and cryptography.

```python
number = 1024

# to_bytes(length, byteorder, signed=False)
# Convert integer to bytes using Big-Endian and Little-Endian order
big_bytes = number.to_bytes(4, byteorder="big")
little_bytes = number.to_bytes(4, byteorder="little")

print("Big-Endian bytes:", big_bytes)
print("Little-Endian bytes:", little_bytes)

# from_bytes(bytes, byteorder, signed=False)
# Convert bytes back to integer
reconstructed = int.from_bytes(big_bytes, byteorder="big")
print("Reconstructed integer:", reconstructed)
```

## 3. Ratio and Duck Typing Attributes
Integers implement methods and attributes to ensure compatibility with `float`, `complex`, and `Fraction` types.

```python
val = 42

# as_integer_ratio(): Returns a (numerator, denominator) pair
print("as_integer_ratio:", val.as_integer_ratio())

# Numeric interoperability attributes:
print("real:", val.real)
print("imag:", val.imag)
print("numerator:", val.numerator)
print("denominator:", val.denominator)
print("conjugate:", val.conjugate())
```

## 4. Complementary Built-in Functions
Python provides built-in functions that work hand-in-hand with integers.

```python
num = 255

# String conversions for different bases
print("bin:", bin(num))  # Binary representation
print("hex:", hex(num))  # Hexadecimal representation
print("oct:", oct(num))  # Octal representation

# Math helpers
quotient, remainder = divmod(100, 7)  # Returns (quotient, remainder)
print(f"divmod(100, 7): quotient={quotient}, remainder={remainder}")

# Efficient modular exponentiation: (base ** exp) % mod
print("pow(2, 10, 1000):", pow(2, 10, 1000))

print("abs(-50):", abs(-50))
```
