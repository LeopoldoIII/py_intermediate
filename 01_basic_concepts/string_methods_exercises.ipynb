## Exercise 2: Password Validator
Write a program that checks if a given password is valid. A valid password must:
- Contain only alphanumeric characters (letters and numbers).
- Not contain any spaces.

```python
password_1 = "SecurePass123"
password_2 = "invalid pass!"

# Write your code here
def validate_password(string: str) -> bool:
    return string.isalnum()

def validate_password_all(string: str) -> bool:
    return all(char.isalnum() for char in string)

def validate_password_for_optimized(string: str) -> bool:
    for i in string:
        if not i.isalnum():
            return False  # If you find ONE that doesn't comply, the whole validation fails
    return True # If the loop finishes without failing, then it is valid

import re
def validate_password_regex(string: str) -> bool:
    # ^ indicates the start of the string
    # [a-zA-Z0-9]+ indicates letters and numbers 1 or more times
    # $ indicates the end of the string
    # fullmatch() ensures that the ENTIRE string matches the pattern
    return bool(re.fullmatch(r'[a-zA-Z0-9]+', string))
```
