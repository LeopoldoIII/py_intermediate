## Section 1: In-Memory Serialization (dumps & loads)

### 🎯 Objective
Understand the difference between `json.dumps()` (object to string) and `json.loads()` (string to object), formatting indentation, and handling key types.

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Encoding and Decoding Dicts
```python
import json

food_ratings = {
    "organic dog food": 2,
    "human food": 10
}

# Serialize to string
json_str = json.dumps(food_ratings, indent=2)
print("JSON string:
", json_str)

# Deserialize back to Python dict
parsed = json.loads(json_str)
print("Parsed rating:", parsed["human food"])  # 10
```

```python
import json

def parse_and_extract_users(payload_json: str, min_age: int) -> list[str]:
    if not payload_json:
        return []
    try:
        users = json.loads(payload_json)
        return sorted([u["name"] for u in users if u.get("age", 0) >= min_age])
    except Exception:
        return []
```
