## Section 1: DOM Navigation & Table Parsing

### 🎯 Objective
Parse HTML strings with `BeautifulSoup`, search elements using CSS selectors (`.select()`, `.find_all()`), and extract text and attributes (`tag['href']`).

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Parsing HTML Elements
```python
from bs4 import BeautifulSoup

html_doc = """
<html>
  <body>
    <div class="user-card" id="u1">
      <h2 class="name">Alice Smith</h2>
      <a href="/profile/alice">Profile</a>
    </div>
  </body>
</html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
name = soup.select_one(".name").get_text(strip=True)
link = soup.select_one("a")["href"]
print(f"Name: {name}, Link: {link}")
```

```python
from bs4 import BeautifulSoup

def extract_html_table_records(html_content: str) -> list[dict[str, str]]:
    if not html_content:
        return []
    soup = BeautifulSoup(html_content, "html.parser")
    table = soup.find("table")
    if not table:
        return []
    headers = [th.get_text(strip=True) for th in table.find_all("th")]
    results = []
    for row in table.find_all("tr"):
        cells = [td.get_text(strip=True) for td in row.find_all("td")]
        if cells and len(cells) == len(headers):
            results.append(dict(zip(headers, cells)))
    return results
```
