### 📖 1. Code Samples: Cleaning Text & Word Counting

#### Example: Stripping Punctuation & Counting Words
Punctuation characters attached to words (e.g. `"hello,"` vs `"hello"`) distort word frequency. Using `str.maketrans` efficiently removes punctuation in a single C-level pass.


````python
import string
from collections import Counter

raw_text = "Python is amazing, and Python is fast! Is it fun? Yes, python is fun."

# Remove punctuation and normalize to lowercase
translator = str.maketrans('', '', string.punctuation)
clean_text = raw_text.translate(translator).lower()
words = clean_text.split()

word_counts = Counter(words)
print("Clean words:", words)
print("Frequencies:", word_counts.most_common(3))
````

```python
import string
from collections import defaultdict


def word_length_frequencies(text: str) -> dict[int, int]:
    translator = str.maketrans("", "", string.punctuation)
    clean_words = text.translate(translator).split()
    freqs = defaultdict(int)
    for word in clean_words:
        freqs[len(word)] += 1
    return dict(freqs)
```
