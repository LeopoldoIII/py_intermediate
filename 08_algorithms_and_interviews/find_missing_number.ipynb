# Finding the Missing Number

### 🎯 Objective
Learn how to solve the classic **Missing Number** interview problem using two optimal strategies:
1. **Mathematical Sum Formula** (Gauss summation: $S = \frac{n(n+1)}{2}$).
2. **Bitwise XOR Operation** (exploiting $x \oplus x = 0$).

Both methods achieve optimal $O(n)$ time complexity and $O(1)$ auxiliary space complexity.


### 📖 1. Code Samples: Mathematical vs. Bitwise Approaches

#### Example 1: Sum Formula (Gauss)
The sum of integers from $1$ to $n$ is $n(n+1)/2$. Subtracting the sum of the array yields the missing number.


````python
def find_missing_gauss(nums: list[int]) -> int:
    n = len(nums) + 1  # Full sequence has length n
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum


print("Missing in [1, 2, 4, 5]:", find_missing_gauss([1, 2, 4, 5]))
````

#### Example 2: Bitwise XOR
Because $a \oplus a = 0$ and $a \oplus 0 = a$, XORing all numbers from $1$ to $n$ with all numbers in the array cancels out all duplicates, leaving only the missing number. This prevents potential integer overflow in languages with fixed integer sizes.


````python
def find_missing_xor(nums: list[int]) -> int:
    n = len(nums) + 1
    xor_all = 0
    for i in range(1, n + 1):
        xor_all ^= i
    for num in nums:
        xor_all ^= num
    return xor_all


print("Missing in [3, 7, 1, 2, 8, 4, 5]:", find_missing_xor([3, 7, 1, 2, 8, 4, 5]))
````
