#### Example: All Unique Validation
Verify if all elements in a list are distinct by comparing length with its set.


````python
def all_unique(lst: list) -> bool:
    return len(lst) == len(set(lst))


print("all_unique([1, 2, 3]):", all_unique([1, 2, 3]))
print("all_unique([1, 2, 2]):", all_unique([1, 2, 2]))
````
