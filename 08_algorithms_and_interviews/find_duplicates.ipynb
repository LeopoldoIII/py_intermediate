### 📖 1. Code Samples: Deduplication & Set Intersections

#### Example 1: Finding Duplicates with a Hash Set ($O(n)$ Time)
Using a single pass and a tracking set, we can record duplicates in linear time.


````python
def find_duplicates(nums: list[int]) -> list[int]:
    seen: set[int] = set()
    duplicates: set[int] = set()
    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)
    return sorted(list(duplicates))


sample = [1, 2, 3, 2, 4, 5, 1, 6, 2]
print("Duplicates found:", find_duplicates(sample))
````

#### Example 2: List Intersection (Set Operators)
Python sets provide fast C-level implementations of intersection via the `&` operator or `.intersection()`.


````python
list1 = [1, 2, 3, 4, 5, 6, 7]
list2 = [5, 6, 7, 8, 9, 10]

# Using set operator &
common = sorted(list(set(list1) & set(list2)))
print("Intersection:", common)
````
