# List Chunking
Splits a list into sublists of size N.


```python
def chunk(lst: list, size: int) -> list:
    return [lst[i:i + size] for i in range(0, len(lst), size)]

chunk([1, 2, 3, 4, 5, 6, 7], 3)
```

# Removing Falsy Values (False, None, 0, "")
Uses `filter` to clean lists quickly.


```python
def remove_falsy(lst):
    return list(filter(bool, lst))

remove_falsy([0, 1, False, 2, "", 3, "a", None])
```

# Transposing Matrices (Rows to Columns)
Useful for two-dimensional arrays using `zip`.


```python
matrix = [['a', 'b'], ['c', 'd'], ['e', 'f']]
transposed = list(zip(*matrix))
transposed
```

# Flattening Lists
Converts a list of lists into a single-level list.


```python
nested_lists = [[1, 2], [3, 4], [5, 6]]
flattened = [item for sublist in nested_lists for item in sublist]
flattened
```

# Finding the Most Frequent Element
Without importing libraries, using `max` and `key`.


```python
lst = [1, 2, 3, 1, 2, 1, 1, 4]
most_frequent = max(set(lst), key=lst.count)
most_frequent
```

# Reversing a List
In-place vs Slicing approach.


```python
lst1 = [1, 2, 3]
lst2 = [4, 5, 6]

lst1.reverse() # In-place (modifies original, returns None)
reversed_lst = lst2[::-1] # Slicing (creates a copy)

print(lst1)
print(reversed_lst)
```
