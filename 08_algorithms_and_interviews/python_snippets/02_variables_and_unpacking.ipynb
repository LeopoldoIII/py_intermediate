# Basic Unpacking and Catching the Rest (*)
Extract the first values and group the remaining ones.


```python
grades = [100, 95, 80, 85, 90, 70]
first, second, *the_rest = grades
print(first, second)
print(the_rest)
```

# Ignoring Values (_)
When a function returns multiple values but you don't need all of them.


```python
coordinates = [15, 20, 35]
x, _, z = coordinates
print(x, z)
```

# Quick Variable Swapping
Without using a temporary variable.


```python
a = 10
b = 20
a, b = b, a
print("a:", a, "b:", b)
```

# Dictionary Unpacking (**)
To merge dictionaries or pass kwargs.


```python
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

merged = {**dict1, **dict2}
print(merged)
```
