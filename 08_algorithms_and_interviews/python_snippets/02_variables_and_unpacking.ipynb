<details>
<summary>🔍 Solution</summary>

```python
def unpack_stats(numbers: list[int]) -> tuple[int, int, float]:
    head, *middle, tail = numbers
    if not middle:
        return (head, tail, 0.0)
    avg = round(sum(middle) / len(middle), 2)
    return (head, tail, avg)
```
