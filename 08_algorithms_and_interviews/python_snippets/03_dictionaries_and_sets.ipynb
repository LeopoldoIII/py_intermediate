```python
from collections import defaultdict


def invert_dict_multivalue(d: dict[str, int]) -> dict[int, list[str]]:
    inverted = defaultdict(list)
    for k, v in d.items():
        inverted[v].append(k)
    return {k: sorted(v) for k, v in inverted.items()}
```
