# Validating Unique Values
Converting to a Set removes duplicates; if the length changes, there were duplicates.


```python
def all_unique(lst):
    return len(lst) == len(set(lst))

print(all_unique([1, 2, 3, 4]))
print(all_unique([1, 1, 2, 3]))
```

# Counting Frequencies (Counter)
The fastest way to count elements in interviews.


```python
from collections import Counter

words = ['cat', 'dog', 'cat', 'bird', 'dog', 'cat']
count = Counter(words)
print(count)
print(count.most_common(1))
```

# Avoiding KeyError with `.get()`
If the key does not exist, it returns None or a default value.


```python
user = {'name': 'John', 'age': 30}

# print(user['last_name']) # Would raise KeyError
print(user.get('last_name')) # Returns None
print(user.get('last_name', 'Not specified')) # Default value
```
