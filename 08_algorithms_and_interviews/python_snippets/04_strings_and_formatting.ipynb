### 📖 1. Reference Code Samples


````python
def is_simple_palindrome(s: str) -> bool:
    clean = s.lower().replace(" ", "")
    return clean == clean[::-1]

print("Is 'racecar' palindrome?", is_simple_palindrome("racecar"))
print("Is 'hello' palindrome?", is_simple_palindrome("hello"))
````

    return clean == clean[::-1]
