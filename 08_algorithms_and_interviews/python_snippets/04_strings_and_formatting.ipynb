# Verifying Palindromes
Compares the string with its reversed version.


```python
def is_palindrome(s):
    s = s.lower().replace(" ", "") # Basic cleanup
    return s == s[::-1]

print(is_palindrome("A man a plan a canal Panama"))
```

# Joining Lists into a String
Much faster than concatenating with '+' in a loop.


```python
words = ["Python", "is", "awesome"]
phrase = " ".join(words)
print(phrase)
```

# Removing Unnecessary Spaces (Strip)
Cleans the start and end of a string.


```python
text = "   hello world   \n"
cleaned = text.strip()
print(f"'{cleaned}'")
```

# Advanced F-Strings Formatting
Aligning text and zero-padding.


```python
number = 42
text = "Hello"

print(f"{number:05}") # Zero-pad to 5 digits
print(f"{text:>10}") # Right-align, 10 spaces
print(f"{text:^10}") # Center in 10 spaces
```
