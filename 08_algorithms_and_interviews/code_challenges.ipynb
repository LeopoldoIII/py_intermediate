```python
# Fibonacci sequence recursive version 1

def fibonacci(n):
    if not n <= 1:
        return fibonacci(n - 1) + fibonacci(n - 2)
    return n


for i in range(10):
    print(fibonacci(i))
```

```python
def fibo_list(n):
    a,b = 0, 1
    seqence = []
    for _ in range(n + 1):
        seqence.append(a)
        a, b = b, a + b
    return seqence


print(fibo_list(6))
```

```python
# Given a string, find the first non-repeating character in it and return its index.
# If it doesn't exist, return -1. # Note: all the input strings are already lowercase.

import collections


def count_unique(word):
    count = collections.Counter(word)
    for index, value in enumerate(word):
        if count[value] == 1:
            return index
    return -1


print(count_unique("alphabet"))
print(count_unique("barbados"))
print(count_unique("crunchy"))
```

Problem: Find the First Non-Repeated Character in a String
Objective: Write a Python function that takes a string as input and returns the first non-repeated character in the string. If all characters are repeated or the string is empty, return None.

Example:

Input: "swiss"
Output: "w" (since 'w' is the first character that does not repeat)

```python
def first_non_repeated(text: str) -> str:
    text = text.lower()
    char_count = {}
    for letter in text:
        letter_count = len([char for char in text if char == letter])
        char_count[letter] = letter_count

    for char in text:
        if char_count[char] == 1:
            return char

    return None


# Test cases
print(first_non_repeated("swiss"))  # Output: 'w'
print(first_non_repeated("programming"))  # Output: 'r'
print(first_non_repeated("aabbcc"))  # Output: None
print(first_non_repeated(""))


# Revised Code:
# Here's an optimized version of your solution:

def first_non_repeated(text: str) -> str:
    text = text.lower()
    char_count = {}

    # First pass: Count each character's occurrences
    for letter in text:
        if letter in char_count:
            char_count[letter] += 1
        else:
            char_count[letter] = 1

    # Second pass: Find the first non-repeated character
    for char in text:
        if char_count[char] == 1:
            return char

    return None


# Test cases
print(first_non_repeated("swiss"))  # Output: 'w'
print(first_non_repeated("programming"))  # Output: 'r'
print(first_non_repeated("aabbcc"))  # Output: None
print(first_non_repeated(""))  # Output: None

```

```python
def is_palindrome(text: str) -> bool:
    # convert to lowercase and remove spaces and punctuation
    filtered_text = ''.join(char.lower() for char in text if char.isalnum())
    # check if filtered_text is equal to it's reverse
    return filtered_text == filtered_text[::-1]


# Test cases
print(is_palindrome("A man a plan a canal Panama"))  # Output: True
print(is_palindrome("Hello"))  # Output: False
print(is_palindrome("Racecar"))  # Output: True
print(is_palindrome("No lemon, no melon"))  # Output: True
print(is_palindrome(""))  # Output: True
```

## Exercise: FizzBuzz with a Twist
Objective: Write a Python function that takes a number n and prints the numbers from 1 to n. But there’s a twist:

* For multiples of 3, print "Fizz" instead of the number.
* For multiples of 5, print "Buzz" instead of the number.
* For numbers that are multiples of both 3 and 5, print "FizzBuzz".
* Otherwise, print the number itself.

```python
def fizzbuzz(number: int) -> None:
    for n in range(1, number + 1):
        if n % 3 == 0 and n % 5 == 0:
            print("FizzBuzz")
        elif n % 3 == 0:
            print("Fizz")
        elif n % 5 == 0:
            print("Buzz")
        else:
            print(n)


print(fizzbuzz(15))
```

### **Exercise: Frequency of Maximum Value in Subsegments (HackerRank Style)**
**Objective:** Given an array of integers `numbers` and an array of queries `q` (which represent 1-based starting indices), for each query:
1. Find the maximum value in the subsegment of `numbers` starting from the given index to the end of the array.
2. Count the frequency (how many times it appears) of that maximum value in the subsegment.
3. Return the results as an array.

*Note: The queries use 1-based indexing, so you will need to adjust them for Python's 0-based indexing.*

**Example:**
- **Input:** `numbers = [5, 4, 5, 3, 2]`, `q = [1, 2, 3, 4, 5]`
- **Output:** `[2, 1, 1, 1, 1]`
- **Explanation:**
  - For query `1` (index 0), subsegment is `[5, 4, 5, 3, 2]`. Max is 5, appears 2 times.
  - For query `2` (index 1), subsegment is `[4, 5, 3, 2]`. Max is 5, appears 1 time.
  - For query `3` (index 2), subsegment is `[5, 3, 2]`. Max is 5, appears 1 time.
  - For query `4` (index 3), subsegment is `[3, 2]`. Max is 3, appears 1 time.
  - For query `5` (index 4), subsegment is `[2]`. Max is 2, appears 1 time.

```python
def frequency_of_max_new_users(numbers, q):
    # 1. Create an empty array to store the results
    result = []
    
    # 2. Iterate through each query in the queries array (q)
    for query in q:
        # 3. Adjust the index (queries use 1-based indexing, Python uses 0-based)
        start_index = query - 1
        
        # 4. Get the subsegment of the list from start_index to the end
        segment = numbers[start_index:]
        
        # 5. Find the maximum value in that segment
        max_value = max(segment)
        
        # 6. Count how many times the maximum value appears in the segment
        times_it_appears = segment.count(max_value)
        
        # 7. Append the frequency to our results list
        result.append(times_it_appears)
        
    # 8. Return the list containing all the answers
    return result


# Example Test Case:
numbers_list = [5, 4, 5, 3, 2]
queries_list = [1, 2, 3, 4, 5]
print(frequency_of_max_new_users(numbers_list, queries_list))
```

### **Exercise: Count Valid Pairs with Difference 'k'**
**Objective:** Given an array of integers `numbers` and an integer `k`, find the number of unique pairs `(a, b)` such that `a + k = b`. 
Duplicates must be removed before counting to ensure that the same pair is not counted multiple times.

**Example:**
- **Input:** `numbers = [1, 1, 2, 2, 3, 3]`, `k = 1`
- **Output:** `2`
- **Explanation:** The unique numbers are `{1, 2, 3}`, and the valid pairs are `(1, 2)` and `(2, 3)`.

```python
def count_pairs(numbers, k):
    print("Numbers:", numbers)
    print("Target difference (k):", k)    
    
    # 1. Use a set to instantly remove duplicates (O(n) time complexity)
    unique_numbers = set(numbers)
            
    result = 0
    
    # 2. Perform the mathematical logic on the cleaned set
    for value in unique_numbers:
        target = value + k 
        # Checking 'in' a set is extremely fast: O(1) time complexity
        if target in unique_numbers:
            result += 1
            
    return result


# Example Test Case:
print("Total valid pairs:", count_pairs([1, 1, 2, 2, 3, 3], 1)) 
# Should output 2, because the unique numbers are {1, 2, 3}, and the valid pairs are (1,2) and (2,3).
```

### Exercise 1: Two Sum Unique Pairs
**Description:**
Given an array of integers `numbers` and an integer `target`, determine the number of unique pairs $(a, b)$ in the list such that $a + b = target$. 
Consider the pairs as distinct if at least one element differs between them. The order within the pair does not matter (i.e., the pair `(a, b)` is considered equal to the pair `(b, a)`).

**Example 1:**
- **Input:** `numbers = [1, 2, 3, 4, 5, 2, 3]`, `target = 5`
- **Output:** `2`
- **Explanation:** The unique pairs that sum to 5 are `(1, 4)` and `(2, 3)`. 

**Example 2:**
- **Input:** `numbers = [2, 2, 2, 2]`, `target = 4`
- **Output:** `1`
- **Explanation:** The only valid pair is `(2, 2)`.

```python
def count_target_pairs(numbers: list[int], target: int) -> int:

    unique_numbers = set(numbers)

    result = set()

    for num in unique_numbers:
        complement = target - num
        if complement in unique_numbers:
            pair_sorted = (min(num, complement), max(num, complement))
            result.add(pair_sorted)
    
    return len(result)

print(count_target_pairs([1, 2, 3, 4, 5, 2, 3], 5)) # Debería imprimir: 2
print(count_target_pairs([2, 2, 2, 2], 4))          # Debería imprimir: 1


        
```
