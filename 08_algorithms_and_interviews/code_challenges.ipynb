def count_pairs_with_diff_k(numbers: list[int], k: int) -> int:

```python
def count_unique_two_sum_pairs(numbers: list[int], target: int) -> int:
    seen = set()
    pairs = set()
    for num in numbers:
        complement = target - num
        if complement in seen:
            pairs.add((min(num, complement), max(num, complement)))
        seen.add(num)
    return len(pairs)
```

```python
def has_zero_sum_subarray(numbers: list[int]) -> bool:
    prefix_sums = set()
    current_sum = 0
    for num in numbers:
        current_sum += num
        if current_sum == 0 or current_sum in prefix_sums:
            return True
        prefix_sums.add(current_sum)
    return False
```

<details>

Use a stack and matching map `{')': '(', '}': '{', ']': '['}`. On open bracket push to stack; on closing bracket check `stack.pop() == map[char]`.

<details>
<summary>🔍 Solution</summary>

```python
def is_valid_parentheses(s: str) -> bool:
    bracket_map = {")": "(", "}": "{", "]": "["}
    stack = []
    for char in s:
        if char in bracket_map:
            top_element = stack.pop() if stack else "#"
            if bracket_map[char] != top_element:
                return False
        else:
            stack.append(char)
    return not stack
```

def get_moon_phase(date_str: str) -> str:

    target_date = datetime.strptime(date_str, "%Y-%m-%d")
    days = (target_date - ref_date).days
