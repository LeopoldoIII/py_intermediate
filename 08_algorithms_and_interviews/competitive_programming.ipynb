Tu tarea: Escribe una función llamada filtrar_repetidos(lista_palabras) que reciba una lista de textos y devuelva una nueva lista que contenga únicamente las palabras que aparecieron más de una vez en la lista original. Además, el resultado no debe tener duplicados.

```python
entrada = ["python", "java", "python", "javascript", "java", "python", "c++"]


def filtrar_repetidos(entrada):
    temp_dic = {}

    for word in entrada:
        if word in temp_dic.keys():
            temp_dic[word] += 1
        else: 
            temp_dic[word] = 1 

    return list({k: v for k, v in  temp_dic.items() if v > 1}.keys())


# Si ejecutamos tu función:
print(filtrar_repetidos(entrada)) 

# Debería regresar exactamente: ["python", "java"]
# (Porque 'python' y 'java' se repiten, pero 'javascript' y 'c++' solo aparecen una vez)
```
