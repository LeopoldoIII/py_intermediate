### 📖 1. Code Samples: Bit Operations and Binary Gap

#### Example: Analyzing Binary Strings
A **binary gap** within a positive integer $N$ is any maximal sequence of consecutive zeros that is surrounded by ones at both ends in the binary representation of $N$.


````python
def binary_gap_demo(n: int) -> int:
    bin_str = f"{n:b}"
    print(f"Number {n} in binary: {bin_str}")
    max_gap = 0
    current_gap = 0
    in_gap = False

    for bit in bin_str:
        if bit == '1':
            if in_gap:
                max_gap = max(max_gap, current_gap)
            in_gap = True
            current_gap = 0
        elif in_gap:
            current_gap += 1

    return max_gap


print("Binary gap for 9 (1001):", binary_gap_demo(9))      # gap of 2
print("Binary gap for 529 (1000010001):", binary_gap_demo(529))  # gap of 4
````

    return max((len(gap) for gap in gaps), default=0)

def single_number(nums: list[int]) -> int:
    result = 0
    for num in nums:
        result ^= num
    return result
