# Two Sum and Pairs That Sum to a Target

### 🎯 Objective
Master the **Two Sum** pattern, the single most popular coding interview question. 

You will analyze the trade-offs between:
- **Brute Force**: $O(n^2)$ time, $O(1)$ space.
- **Hash Table / Set (Complement Lookup)**: $O(n)$ time, $O(n)$ space.
- **Two Pointers (Sorted Array)**: $O(n \log n)$ time, $O(1)$ space.


### 📖 1. Code Samples: Brute Force vs. Hash Table

#### The Brute Force Approach ($O(n^2)$)
Examine all possible pairs with nested loops.


````python
def find_pairs_brute_force(nums: list[int], target: int) -> list[tuple[int, int]]:
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.append((nums[i], nums[j]))
    return pairs


print("Pairs:", find_pairs_brute_force([2, 7, 4, 5, 1, 3, 6, 8], 9))
````

#### The Optimal Hash Set Approach ($O(n)$)
For each element `num`, check if its complement `target - num` was already visited in our set.


````python
def find_pairs_hash_set(nums: list[int], target: int) -> list[tuple[int, int]]:
    pairs = []
    seen = set()
    for num in nums:
        complement = target - num
        if complement in seen:
            pairs.append((min(num, complement), max(num, complement)))
        seen.add(num)
    return sorted(list(set(pairs)))


print("Optimized pairs:", find_pairs_hash_set([2, 7, 4, 5, 1, 3, 6, 8], 9))
````

    return None
