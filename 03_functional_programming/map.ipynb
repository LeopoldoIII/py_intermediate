# Functional Programming: The map() Function

The built-in `map(function, iterable, ...)` applies a transformation function to every item of an iterable and yields an iterator of results.

This notebook demonstrates the functional `map()` pattern alongside idiomatic list comprehensions, preserving real-world use cases (arithmetic, temperature conversions, type casting, and structured record transformations).

## Section 1: Number Transformations (Map vs Comprehension)

### 🎯 Objective
Understand how `map()` lazily applies a unary function to numerical collections, and contrast it with list comprehensions.

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Squaring Numbers
```python
numbers = [1, 2, 3, 4, 5]

# Approach 1: Functional map() with lambda
squared_map = list(map(lambda x: x ** 2, numbers))

# Approach 2: List comprehension equivalent
squared_comp = [x ** 2 for x in numbers]

print("map() squared:         ", squared_map)   # [1, 4, 9, 16, 25]
print("comprehension squared: ", squared_comp)  # [1, 4, 9, 16, 25]
```

## Section 4: Complex Dictionary Transformations & Pricing

    return list(map(
        lambda item: {
            **item,
            "tax": round(item["price"] * tax_rate, 2),
            "total": round(item["price"] + round(item["price"] * tax_rate, 2), 2)
        },
        items
    ))
