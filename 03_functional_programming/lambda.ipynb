# Lambda Functions in Python

A `lambda` function is a small, anonymous function defined with the `lambda arguments: expression` syntax. It can take any number of arguments, but can only have one expression.

## Section 3: Custom Sorting Keys & Alphanumeric Natural Ordering

### 🎯 Objective
Use lambdas as the `key` parameter in `sorted()` to parse complex string identifiers (e.g. `['id1', 'id2', 'id30', 'id3']`).

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Lexicographical vs Numeric Suffix Sort
```python
ids = ['id1', 'id2', 'id30', 'id3', 'id22', 'id100']

# Default Lexicographical sort (fails numerically: 'id100' before 'id2'):
print("Lexicographical:", sorted(ids))
# ['id1', 'id100', 'id2', 'id22', 'id3', 'id30']

# Natural numeric sort using lambda:
print("Natural sort:   ", sorted(ids, key=lambda s: int(s[2:])))
# ['id1', 'id2', 'id3', 'id22', 'id30', 'id100']
```
