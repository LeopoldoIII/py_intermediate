# Functional Programming: The reduce() Function

The `functools.reduce(function, iterable[, initializer])` function applies a function of two arguments cumulatively to the items of a sequence, from left to right, reducing the iterable to a single cumulative value.

#### Demonstration A: Summing Numbers with reduce()
```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Using reduce
sum_reduce = reduce(lambda acc, x: acc + x, numbers)

# Pythonic built-in
sum_builtin = sum(numbers)

print("reduce sum: ", sum_reduce)   # 15
print("builtin sum:", sum_builtin)  # 15
```

```python
from functools import reduce
from typing import Any

def aggregate_order_totals(orders: list[dict[str, Any]]) -> dict[str, float]:
    subtotal = reduce(lambda acc, item: acc + (item["price"] * item["quantity"]), orders, 0.0)
    tax = round(subtotal * 0.1, 2)
    total = round(subtotal + tax, 2)
    return {"subtotal": round(subtotal, 2), "tax": tax, "total": total}
```
