```python
def parse_cli_flags(args: list[str]) -> dict[str, str]:
    flags = {}
    if len(args) <= 1:
        return flags
    for arg in args[1:]:
        if arg.startswith("--") and "=" in arg:
            key, val = arg[2:].split("=", 1)
            flags[key] = val
    return flags
```
