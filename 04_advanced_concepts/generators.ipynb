# Advanced Python: Generators & Iterators

Generators are functions that return an iterator yielding values on demand using `yield` rather than returning an entire in-memory collection. This provides massive memory efficiency when processing large or infinite streams.

## Section 1: The yield Statement & Lazy Evaluation

### 🎯 Objective
Understand how `yield` pauses function state between iterations, saving execution context and allowing progressive value generation.

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Countdown Generator
```python
def count_down(start: int):
    current = start
    while current > 0:
        yield current
        current -= 1

gen = count_down(3)
print("First:", next(gen))   # 3
print("Second:", next(gen))  # 2
print("Third:", next(gen))   # 1
```

## Section 2: Batching & Chunk Processing

### 🎯 Objective
Process large collections in fixed-size chunks to minimize RAM consumption in batch processing pipelines.

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Generator Expression Memory Profile
```python
import sys

# List comprehension: full allocation upfront
full_list = [x ** 2 for x in range(10000)]
# Generator expression: lazy evaluation
gen_exp = (x ** 2 for x in range(10000))

print("List memory:     ", sys.getsizeof(full_list), "bytes")
print("Generator memory:", sys.getsizeof(gen_exp), "bytes")  # Dramatically smaller!
```

```python
from typing import Any, Iterator

def batch_stream(items: list[Any], batch_size: int) -> Iterator[list[Any]]:
    if batch_size <= 0:
        return
    for i in range(0, len(items), batch_size):
        yield items[i:i + batch_size]
```
