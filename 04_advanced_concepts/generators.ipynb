# Generators in Python

Generators are a simple and powerful way to create iterators. Unlike standard functions that compute and return a whole list of values at once, generators use the `yield` keyword to return one value at a time, pausing their state between each call. This makes them extremely memory efficient for large datasets.

```python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

# Using the generator
counter = count_up_to(5)
print(next(counter))
print(next(counter))
```

### Generator Expressions
Similar to list comprehensions, you can create generators using parentheses `()` instead of square brackets `[]`.

```python
# This creates a generator object, not a full list in memory
gen_exp = (x**2 for x in range(10))

for num in gen_exp:
    print(num, end=" ")
```
