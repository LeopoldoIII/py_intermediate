## Section 2: Stateful Running Averages

### 🎯 Objective
Maintain a cumulative statistical history across successive calls without using global variables or class boilerplate.

---

### 📖 1. Code Samples (Comparative Demonstrations)

#### Demonstration A: Cumulative String Builder
```python
def make_accumulator():
    history = []
    def record(val: str) -> list[str]:
        history.append(val)
        return list(history)
    return record

acc = make_accumulator()
print("1st:", acc("log1"))  # ['log1']
print("2nd:", acc("log2"))  # ['log1', 'log2']
```

```python
from typing import Callable

def make_running_average() -> Callable[[float], float]:
    total = 0.0
    count = 0
    def add_number(val: float) -> float:
        nonlocal total, count
        total += val
        count += 1
        return round(total / count, 2)
    return add_number
```
